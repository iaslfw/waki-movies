# 1. Einleitung und Projektziel

## 1.1 Ausgangssituation

Filmempfehlungen sind ein typisches Anwendungsfeld für Empfehlungssysteme. Viele bekannte Plattformen schlagen Inhalte auf Basis von Genres, Bewertungen, Nutzerverhalten oder ähnlichen Filmen vor. Solche Ansätze sind grundsätzlich hilfreich, bilden aber nicht immer die tatsächliche Suchabsicht eines Nutzers ab. Häufig sucht ein Nutzer nicht nach einem festen Genre, sondern nach einer bestimmten Stimmung, Atmosphäre oder inhaltlichen Richtung. Eine Anfrage kann zum Beispiel lauten: „Ich möchte einen spannenden Film sehen, der nicht zu düster ist“ oder „Ich suche etwas Lustiges mit Abenteuergefühl“.

Klassische Genre-Kategorien sind für solche Suchanfragen nur eingeschränkt geeignet. Ein Film kann mehreren Stimmungen gleichzeitig entsprechen. Ein Science-Fiction-Film kann beispielsweise düster, spannend, emotional oder humorvoll sein. Ebenso kann ein Drama gleichzeitig melancholisch, romantisch und hoffnungsvoll wirken. Für ein Empfehlungssystem ist es daher sinnvoll, Filme nicht nur über starre Kategorien zu betrachten, sondern über feinere Merkmale wie Stimmung, Atmosphäre, Tempo oder thematische Schwerpunkte.

## 1.2 Ziel des Projekts

Ziel des Projekts „WaKi-Movies“ ist die Entwicklung eines KI-gestützten Empfehlungssystems für Filme. Das System soll eine freie Texteingabe eines Nutzers verarbeiten und daraus passende Filmempfehlungen ableiten. Im Mittelpunkt steht dabei nicht die reine Genre-Suche, sondern die Vorhersage sogenannter Mood-Tags. Diese Mood-Tags beschreiben latente Eigenschaften eines Films, zum Beispiel ob ein Film spannend, lustig, düster, emotional oder abenteuerlich wirkt.

Das Projekt kombiniert Methoden aus Natural Language Processing, Deep Learning und Empfehlungssystemen. Eine Nutzereingabe wird zunächst von einem trainierten Sprachmodell analysiert. Anschließend erzeugt das Modell einen Vektor, der die vorhergesagten Stimmungen der Eingabe abbildet. Dieser Nutzervektor wird mit den gespeicherten Vektoren der Filme verglichen. Die Filme mit der höchsten Ähnlichkeit werden anschließend als Empfehlung ausgegeben.

## 1.3 Grundidee des Movie Mood-Predictors

Die Grundidee besteht aus zwei zentralen Schritten. Im ersten Schritt wird die Freitexteingabe des Nutzers in einen Mood-Vektor umgewandelt. Dafür wird ein Transformer-basiertes Sprachmodell verwendet, das für eine Multi-Label-Klassifikation trainiert wurde. Multi-Label bedeutet in diesem Zusammenhang, dass nicht nur eine einzelne Klasse vorhergesagt wird. Stattdessen können mehrere Stimmungen gleichzeitig erkannt werden.

Im zweiten Schritt wird dieser vorhergesagte Mood-Vektor mit den vorhandenen Filmvektoren verglichen. Jeder Film besitzt ebenfalls einen Vektor, der aus den zugrunde liegenden Tag-Daten erzeugt wurde. Für den Vergleich wird die Cosine Similarity verwendet. Dadurch kann gemessen werden, wie ähnlich die Stimmung der Nutzereingabe zur Stimmung eines Films ist. Die Filme mit der höchsten Ähnlichkeit werden als Top-Empfehlungen zurückgegeben.

Die Anwendung wird über einen Telegram-Bot nutzbar gemacht. Nutzer müssen daher nicht direkt mit dem Python-Code arbeiten. Stattdessen können sie ihre Filmwünsche als normale Chatnachricht eingeben. Der Bot verarbeitet die Anfrage, ruft die Vorhersage des Modells auf, berechnet passende Empfehlungen und gibt diese anschließend im Chat aus.

# 2. Theoretische Grundlagen

## 2.1 Empfehlungssysteme

Empfehlungssysteme dienen dazu, Nutzern aus einer großen Menge an möglichen Objekten passende Vorschläge zu machen. Im Filmbereich ist dies besonders relevant, da die Anzahl verfügbarer Filme und Serien sehr groß ist und Nutzer aus dieser Menge passende Inhalte auswählen müssen. Empfehlungssysteme können diesen Auswahlprozess unterstützen, indem sie Merkmale, Präferenzen oder Ähnlichkeiten zwischen Inhalten berücksichtigen [1, S. 1].

Klassische Filmempfehlungen basieren häufig auf Genres, Bewertungen oder ähnlichem Nutzerverhalten. Solche Ansätze sind grundsätzlich nützlich, bilden aber nicht immer die konkrete Suchabsicht eines Nutzers ab. Häufig sucht ein Nutzer nicht nach einem festen Genre, sondern nach einer bestimmten Stimmung, Atmosphäre oder Wirkung. Sarhan et al. beschreiben in diesem Zusammenhang, dass moderne Filmempfehlungssysteme auch emotionale Zustände und individuelle Präferenzen berücksichtigen können, um passendere Empfehlungen zu erzeugen [1, S. 1].

Das Projekt „WaKi-Movies“ greift diesen Gedanken auf. Die Empfehlung erfolgt nicht ausschließlich über feste Genres, sondern über sogenannte Mood-Tags. Diese Tags beschreiben feinere Eigenschaften eines Films, zum Beispiel Spannung, Humor, düstere Atmosphäre oder emotionale Wirkung. Dadurch wird ein Film nicht nur einer Kategorie zugeordnet, sondern durch mehrere Merkmale beschrieben.

Das Projekt kann als inhaltsbasiertes Empfehlungssystem eingeordnet werden. Die Empfehlung entsteht durch den Vergleich der Nutzereingabe mit vorhandenen Filmeigenschaften. Jeder Film wird durch einen Vektor beschrieben. Dieser Vektor enthält Informationen darüber, welche Mood-Tags auf den Film zutreffen. Die Nutzereingabe wird ebenfalls in einen solchen Vektor überführt. Anschließend werden beide Vektoren miteinander verglichen.

## 2.2 Natural Language Processing

Natural Language Processing beschreibt Verfahren zur maschinellen Verarbeitung natürlicher Sprache. Ziel ist es, geschriebene oder gesprochene Sprache so aufzubereiten, dass sie von Computersystemen analysiert und weiterverarbeitet werden kann. Pröhl et al. beschreiben NLP als ein Anwendungsfeld künstlicher Intelligenz, das unter anderem zur Automatisierung und Optimierung von Prozessen eingesetzt werden kann [2, S. 1094–1095].

Für das Projekt ist NLP zentral, weil die Eingabe des Nutzers nicht aus festen Auswahlfeldern besteht, sondern als freie Texteingabe erfolgt. Ein Satz wie „I want a funny but adventurous movie“ enthält für Menschen eine gut verständliche Bedeutung. Für ein KI-System muss dieser Text jedoch zuerst in eine verarbeitbare Form überführt werden. Dazu werden Texte typischerweise in kleinere Einheiten zerlegt, numerisch kodiert und anschließend durch ein Modell verarbeitet.

Moderne Text-Mining-Verfahren basieren zunehmend auf maschinellem Lernen und neuronalen Netzen. Biemann et al. beschreiben, dass heute insbesondere maschinelles Lernen und neuronale Netze eine wichtige Grundlage für die automatische Verarbeitung von Text bilden [3, S. V–VI]. Dieser Grundgedanke wird auch im Projekt „WaKi-Movies“ genutzt. Die Nutzereingabe wird durch einen Tokenizer verarbeitet und anschließend an ein trainiertes Sprachmodell übergeben. Dadurch entsteht aus einer normalen Chatnachricht eine maschinell auswertbare Repräsentation.

Für das Projekt ist außerdem relevant, dass NLP nicht nur einzelne Wörter betrachtet. Entscheidend ist der Zusammenhang innerhalb des Satzes. Die Aussage „funny but adventurous“ enthält beispielsweise mehrere Bedeutungsaspekte. Ein geeignetes Sprachmodell muss daher erfassen können, dass eine Eingabe mehrere gewünschte Eigenschaften gleichzeitig beschreibt.

## 2.3 Transformer, DistilBERT und Fine-Tuning

Für die Verarbeitung der Texteingaben wird im Projekt ein Transformer-basiertes Sprachmodell verwendet. Transformer-Modelle haben sich in vielen NLP-Aufgaben etabliert, weil sie Zusammenhänge innerhalb eines Textes über Attention-Mechanismen erfassen können. Dadurch kann ein Modell nicht nur einzelne Wörter isoliert betrachten, sondern auch deren Kontext im Satz berücksichtigen. Lanquillon und Schacht ordnen Transformer-Modelle und Attention-Mechanismen als zentrale Bausteine moderner Deep-Learning-Architekturen ein [4, S. 138–155].

Auch Biemann et al. beschreiben BERT als Modell auf Basis der Transformer-Architektur. Dabei wird der Attention-Mechanismus genutzt, um den Informationsfluss zwischen Wörtern eines Satzes zu modellieren und kontextabhängige Repräsentationen zu erzeugen [3, S. 225]. Genau diese Eigenschaft ist für das Projekt relevant, da die Bedeutung einer Nutzereingabe nicht nur aus einzelnen Schlüsselwörtern besteht, sondern aus dem gesamten sprachlichen Zusammenhang.

Im Projekt wird konkret `distilbert-base-uncased` verwendet. DistilBERT ist eine kompaktere Variante von BERT. Das Modell ist bereits auf großen Textmengen vortrainiert und kann anschließend für eine konkrete Aufgabe weitertrainiert werden. Dieses Vorgehen wird als Fine-Tuning bezeichnet. Dabei wird kein Modell vollständig von Grund auf neu trainiert. Stattdessen wird ein bereits vortrainiertes Modell an eine spezifische Aufgabe angepasst.

Für „WaKi-Movies“ bedeutet das: Das Modell bringt bereits allgemeine Sprachrepräsentationen aus dem Vortraining mit. Durch das Training auf den Filmbeschreibungen und Mood-Labels wird es anschließend auf die konkrete Aufgabe der Mood-Tag-Vorhersage angepasst. Diese Vorgehensweise ist besonders sinnvoll, weil für das Projekt zwar ein eigener Datensatz vorhanden ist, dieser aber nicht mit den Datenmengen vergleichbar ist, die für ein vollständiges Vortraining eines großen Sprachmodells erforderlich wären.

## 2.4 Multi-Label-Klassifikation

Bei einer klassischen Multi-Class-Klassifikation wird genau eine Klasse aus mehreren möglichen Klassen ausgewählt. Für dieses Projekt wäre ein solcher Ansatz ungeeignet. Ein Film kann nicht nur eine einzige Stimmung besitzen. Ein Film kann gleichzeitig spannend, emotional, düster und abenteuerlich sein. Ebenso kann eine Nutzereingabe mehrere gewünschte Eigenschaften enthalten.

Deshalb wird im Projekt eine Multi-Label-Klassifikation verwendet. Lanquillon und Schacht beschreiben Multi-Label-Klassifikation als ein Klassifikationsproblem, bei dem ein Objekt mehreren vorgegebenen Klassen gleichzeitig zugeordnet werden kann [4, S. 87]. Genau diese Eigenschaft ist für „WaKi-Movies“ wichtig. Die Ausgabe des Modells ist kein einzelnes Label, sondern ein Vektor mit Wahrscheinlichkeiten für alle vorhandenen Mood-Tags.

Diese Vorgehensweise passt zur Zielsetzung des Projekts. Die Empfehlung soll nicht auf einer einzigen Kategorie beruhen. Stattdessen soll die Stimmung einer Eingabe differenziert abgebildet werden. Wenn ein Nutzer beispielsweise einen lustigen, aber gleichzeitig abenteuerlichen Film sucht, sollen beide Aspekte in der Modellvorhersage berücksichtigt werden können.

Der entstehende Mood-Vektor bildet anschließend die Grundlage für den Vergleich mit den Filmvektoren. Dadurch wird die Klassifikationsaufgabe direkt mit der späteren Empfehlungslogik verbunden. Das Modell erzeugt also nicht direkt eine Filmempfehlung, sondern zuerst eine strukturierte Darstellung der gesuchten Stimmung.

## 2.5 Sigmoid-Funktion und Wahrscheinlichkeitswerte

Damit aus den Modellwerten interpretierbare Wahrscheinlichkeiten entstehen, wird die Sigmoid-Funktion verwendet. Das Modell gibt zunächst sogenannte Logits aus. Diese Rohwerte können beliebige positive oder negative Werte annehmen. Durch die Sigmoid-Funktion werden sie in Werte zwischen 0 und 1 umgerechnet.

Für Multi-Label-Klassifikation ist diese Vorgehensweise sinnvoll, weil jedes Label unabhängig bewertet wird. Lanquillon und Schacht beschreiben für binäre Klassifikation, dass die logistische Funktion Werte zwischen 0 und 1 liefert und der Ausgabewert als Konfidenz- oder Wahrscheinlichkeitswert interpretiert werden kann [4, S. 87]. Übertragen auf eine Multi-Label-Klassifikation wird diese Logik für jedes Label separat angewendet. Jedes Mood-Tag erhält dadurch eine eigene Wahrscheinlichkeit.

Im Unterschied zu einer Softmax-Ausgabe wird nicht erzwungen, dass nur eine Klasse dominiert. Mehrere Mood-Tags können gleichzeitig hohe Werte erhalten. Das passt zur Projektlogik, weil Filme und Nutzereingaben mehrere Stimmungen gleichzeitig enthalten können. Ein hoher Wert bei einem Mood-Tag bedeutet, dass dieses Tag wahrscheinlich zur Eingabe passt. Ein niedriger Wert bedeutet, dass dieses Tag eher nicht passt.

Im Projekt wird diese Logik sowohl beim Training als auch bei der späteren Vorhersage genutzt. Bei der Inferenz entsteht aus einer Nutzereingabe ein Wahrscheinlichkeitsvektor. Dieser Vektor beschreibt, welche Stimmungen das Modell in der Eingabe erkennt. Die konkrete Schwellenwertentscheidung und die Bewertung der Modellleistung werden im späteren Kapitel zum Modelltraining genauer erläutert.

## 2.6 Cosine Similarity

Nach der Vorhersage der Mood-Tags muss entschieden werden, welche Filme am besten zur Nutzereingabe passen. Dafür wird die Cosine Similarity verwendet. Sie misst die Ähnlichkeit zwischen zwei Vektoren anhand des Winkels zwischen ihnen. Je ähnlicher die Richtung der Vektoren ist, desto höher ist der Ähnlichkeitswert. Dadurch eignet sich die Methode besonders für den Vergleich von Merkmalsvektoren.

Sarhan et al. beschreiben Cosine Similarity als Maß zur Bestimmung der Ähnlichkeit von Datenobjekten unabhängig von ihrer Größe. Die Ähnlichkeit wird dabei über den Kosinus des Winkels zwischen zwei Vektoren berechnet [1, S. 16–17]. Für „WaKi-Movies“ ist diese Eigenschaft relevant, weil sowohl die Nutzereingabe als auch die Filme als Vektoren dargestellt werden.

Im Projekt wird der vorhergesagte Nutzervektor mit den gespeicherten Filmvektoren verglichen. Jeder Film besitzt einen eigenen Mood-Vektor. Die Cosine Similarity berechnet für jeden Film, wie stark dieser Film zur vorhergesagten Stimmung des Nutzers passt. Anschließend werden die Filme nach ihrem Ähnlichkeitswert sortiert. Die Filme mit den höchsten Werten werden als Top-Empfehlungen ausgegeben.

Damit verbindet die Cosine Similarity die Modellvorhersage mit der eigentlichen Empfehlungsfunktion. Das Sprachmodell erkennt zunächst die Stimmung der Eingabe. Der Ähnlichkeitsvergleich bestimmt danach, welche Filme diese Stimmung am besten abbilden.

# 3. Datenbasis und Datenaufbereitung

## 3.1 Verwendete Datenquellen

Für das Projekt werden Filminformationen und Tag-Daten miteinander kombiniert. Die Filminformationen stammen aus einem TMDB-Datensatz. Dieser enthält unter anderem Filmtitel und Filmbeschreibungen. Die Beschreibungen sind für das Projekt besonders wichtig, da sie als Texteingabe für das Modelltraining verwendet werden.

Zusätzlich werden MovieLens-Daten genutzt. MovieLens stellt neben Bewertungsdaten auch sogenannte Genome-Tags bereit. Diese Tags beschreiben Filme über viele unterschiedliche Merkmale. Die Tags gehen über klassische Genres hinaus und können Eigenschaften wie Stimmung, Thema, Atmosphäre oder Stil eines Films abbilden. Dadurch eignen sie sich als Grundlage für die im Projekt verwendeten Mood-Tags.

Die Kombination beider Datenquellen ist für das Projekt zentral. TMDB liefert die textuelle Beschreibung der Filme. MovieLens liefert die strukturierte Tag-Information. Aus beiden Bestandteilen entsteht ein Datensatz, der sowohl Texteingaben als auch zugehörige Label-Vektoren enthält. Genau diese Struktur wird für das Training einer Multi-Label-Klassifikation benötigt.

Filmempfehlungssysteme nutzen häufig unterschiedliche Datenquellen, um bessere Empfehlungen erzeugen zu können. Sarhan et al. zeigen im Kontext von Filmempfehlungen ebenfalls, dass Filmdaten, Nutzerinformationen und textbezogene Informationen für Empfehlungssysteme kombiniert werden können [1, S. 1–2]. Für „WaKi-Movies“ liegt der Fokus dabei nicht auf Nutzerhistorien, sondern auf der Verbindung von Filmbeschreibungen und inhaltlichen Tag-Vektoren.

## 3.2 Bereinigung der Filmdaten

Vor dem Training müssen die Rohdaten bereinigt werden. Im Projekt werden zunächst nur die für das Modell relevanten Spalten aus dem TMDB-Datensatz verwendet. Dazu gehören die Film-ID, der Titel und die Filmbeschreibung. Filme ohne Beschreibung werden entfernt, da sie nicht sinnvoll als Texteingabe für das Sprachmodell genutzt werden können.

Anschließend werden die TMDB-Daten mit den MovieLens-Links verbunden. Die Link-Datei stellt die Verbindung zwischen TMDB-IDs und MovieLens-IDs her. Diese Zuordnung ist notwendig, weil die Filmbeschreibungen aus TMDB stammen, während die Tag-Informationen aus MovieLens bzw. den Genome-Daten stammen. Ohne diese Verbindung könnten die Texte nicht eindeutig den passenden Tag-Vektoren zugeordnet werden.

Die Bereinigung ist ein wichtiger Schritt in der Datenpipeline. Textdaten liegen häufig nicht direkt in einer Form vor, die für maschinelles Lernen geeignet ist. Biemann et al. beschreiben, dass bei der maschinellen Verarbeitung von Texten typischerweise mehrere Vorverarbeitungsschritte notwendig sind, bevor eine weitere Analyse oder Modellierung erfolgen kann [3, S. 85–90]. Im Projekt betrifft dies vor allem die Auswahl relevanter Spalten, das Entfernen unvollständiger Datensätze und das Zusammenführen der Datenquellen.

## 3.3 Bildung der Mood-Tags

Die Mood-Tags werden aus den Genome-Tag-Daten erzeugt. Die Genome-Daten enthalten Relevance-Werte. Diese Werte geben an, wie stark ein bestimmtes Tag zu einem Film passt. Da das Modell jedoch eine Multi-Label-Klassifikation lernen soll, müssen aus diesen Relevance-Werten Label-Vektoren gebildet werden.

Im Projekt werden zunächst die relevantesten Tags ausgewählt. Dafür werden nur Tags berücksichtigt, deren Relevance-Wert mindestens dem konfigurierten Entscheidungsschwellwert entspricht. Aus diesen Tags werden anschließend die 100 am häufigsten relevanten Tags ausgewählt. Dadurch wird die Label-Dimension begrenzt und das Modell muss nicht auf eine zu große Menge an selten auftretenden Tags trainiert werden.

Im nächsten Schritt wird aus den Tag-Daten eine Matrix gebildet. Jede Zeile steht für einen Film. Jede Spalte steht für ein ausgewähltes Mood-Tag. Der jeweilige Wert gibt an, ob das Tag für diesen Film als relevant gilt. Die Relevance-Werte werden dabei binarisiert. Ein Wert oberhalb des Schwellwerts wird als `1` interpretiert. Ein Wert unterhalb des Schwellwerts wird als `0` interpretiert.

Dadurch entsteht für jeden Film ein Label-Vektor. Dieser Vektor beschreibt, welche Mood-Tags auf den Film zutreffen. Ein solcher Vektor ist die Grundlage für die spätere Multi-Label-Klassifikation. Das Modell lernt also nicht, einem Film genau eine Klasse zuzuordnen. Stattdessen lernt es, auf Basis der Filmbeschreibung mehrere passende Mood-Tags gleichzeitig vorherzusagen.

## 3.4 Erstellung des finalen Datensatzes

Nach der Bereinigung und Tag-Bildung werden die Filminformationen mit den Mood-Vektoren zusammengeführt. Der finale Datensatz enthält drei zentrale Bestandteile: den Filmtitel, die Filmbeschreibung und den zugehörigen Label-Vektor. Die Filmbeschreibung wird dabei als `text` gespeichert. Die Mood-Tags werden als Liste in der Spalte `labels` abgelegt.

Diese Struktur ist für das spätere Training entscheidend. Der Text dient als Eingabe für das Sprachmodell. Der Label-Vektor dient als Zielwert. Das Modell soll lernen, aus einer Filmbeschreibung die passenden Mood-Tags vorherzusagen. Dadurch wird die Verbindung zwischen natürlicher Sprache und strukturierter Stimmungsklassifikation hergestellt.

Der finale Datensatz wird lokal als CSV-Datei gespeichert. Zusätzlich werden die ausgewählten Mood-Tags in einer JSON-Datei gespeichert. Diese JSON-Datei ist wichtig, damit beim Training und bei der späteren Inferenz dieselbe Reihenfolge der Labels verwendet wird. Nur wenn die Reihenfolge der Mood-Tags konsistent bleibt, können die vorhergesagten Wahrscheinlichkeiten korrekt interpretiert und mit den Filmvektoren verglichen werden.

Die Datenaufbereitung bildet damit die Grundlage des gesamten Systems. Ohne die Zusammenführung von Filmbeschreibungen und Tag-Vektoren könnte das Modell keine stimmungsbezogene Textklassifikation lernen. Gleichzeitig stellt die Aufbereitung sicher, dass die späteren Empfehlungen auf denselben Mood-Tags basieren, die auch beim Training verwendet wurden.

# 4. Systemarchitektur

## 4.1 Gesamtprozess

Die Systemarchitektur von „WaKi-Movies“ ist als Pipeline aufgebaut. Ziel ist es, eine freie Nutzereingabe in passende Filmempfehlungen zu überführen. Der Ablauf verbindet die Verarbeitung natürlicher Sprache, die Vorhersage von Mood-Tags, einen Ähnlichkeitsvergleich und die Ausgabe über einen Telegram-Bot.

Der Prozess beginnt mit einer Texteingabe des Nutzers. Diese Eingabe kann eine Stimmung, ein Genre, eine kurze Beschreibung oder einen allgemeinen Filmwunsch enthalten. Die Eingabe wird zunächst tokenisiert und anschließend durch das trainierte DistilBERT-Modell verarbeitet. Das Modell erzeugt daraus einen Mood-Vektor. Dieser Vektor enthält Wahrscheinlichkeitswerte für die verwendeten Mood-Tags.

Danach wird der vorhergesagte Mood-Vektor mit den gespeicherten Filmvektoren verglichen. Jeder Film besitzt einen eigenen Label-Vektor, der aus den MovieLens-Tag-Daten erzeugt wurde. Der Vergleich erfolgt über die Cosine Similarity. Die Filme mit der höchsten Ähnlichkeit werden sortiert und als Top-Empfehlungen im Telegram-Bot ausgegeben.

Der Ablauf lässt sich vereinfacht so darstellen:

```text
Nutzereingabe
    ↓
Tokenisierung
    ↓
Vorhersage durch DistilBERT
    ↓
Mood-Vektor
    ↓
Vergleich mit Filmvektoren
    ↓
Cosine Similarity
    ↓
Top-Empfehlungen
    ↓
Ausgabe im Telegram-Bot
```

## 4.2 Projektstruktur

Das Projekt ist modular aufgebaut. Die wichtigsten Dateien und Ordner sind wie folgt strukturiert:

```text
waki-movies/
├── main.py
├── main.ipynb
├── pyproject.toml
├── uv.lock
├── .env.template
└── src/
    ├── settings.py
    ├── data/
    │   ├── prepare_data.py
    │   └── hugging_face.py
    ├── training/
    │   ├── dataset.py
    │   └── train_model.py
    ├── inference/
    │   ├── inference.py
    │   └── recommender.py
    └── telegram_bot/
        └── bot.py
```

Die Datei `main.py` ist der zentrale Einstiegspunkt der Anwendung. Beim Start wird geprüft, ob der vorbereitete Datensatz, die Mood-Tags und das trainierte Modell lokal vorhanden sind. Falls Daten oder Modell fehlen, werden diese vorbereitet oder von Hugging Face heruntergeladen. Danach werden der `MoodPredictor`, der `MovieRecommender` und der `TelegramBot` initialisiert.

Die Datei `settings.py` enthält zentrale Einstellungen des Projekts. Dazu gehören Pfade, Modellname, Schwellwerte, maximale Eingabelänge und Umgebungsvariablen. Dadurch können wichtige Parameter an einer Stelle verwaltet werden.

Im Ordner `src/data/` liegt die Datenlogik. `prepare_data.py` lädt und verarbeitet die TMDB- und MovieLens-Daten, erzeugt Mood-Label-Vektoren und speichert den finalen Datensatz. `hugging_face.py` enthält Funktionen zum Hochladen oder Herunterladen von Daten und Modellen über Hugging Face.

Der Ordner `src/training/` enthält die Bestandteile für das Modelltraining. In `dataset.py` werden die Trainingsdaten geladen, aufgeteilt und tokenisiert. In `train_model.py` wird das DistilBERT-Modell für die Multi-Label-Klassifikation trainiert und bewertet.

Im Ordner `src/inference/` befindet sich die Logik für die spätere Anwendung des Modells. `inference.py` lädt das trainierte Modell und erzeugt aus einer Nutzereingabe einen Mood-Vektor. `recommender.py` vergleicht diesen Vektor mit den gespeicherten Filmvektoren und gibt die passendsten Filme zurück.

Der Ordner `src/telegram_bot/` enthält die Nutzerschnittstelle. `bot.py` definiert die Telegram-Befehle, verarbeitet Nutzernachrichten und gibt die berechneten Empfehlungen formatiert zurück.

## 4.3 Konfiguration und Umgebungsvariablen

Die Konfiguration wird zentral über `settings.py` und eine Umgebungsdatei verwaltet. Dadurch müssen Pfade, Modellparameter und Tokens nicht direkt im Code verteilt werden.

Die Datei `.env.template` dient als Vorlage für benötigte Umgebungsvariablen. Dazu gehören beispielsweise der Telegram-Bot-Token und ein Hugging-Face-Token. Die echte `.env`-Datei sollte nicht veröffentlicht oder abgegeben werden, da sie sensible Zugangsdaten enthalten kann.

Diese Trennung verbessert die Übersichtlichkeit und erleichtert die Reproduzierbarkeit. Andere Nutzer können anhand der Vorlage erkennen, welche Variablen benötigt werden, ohne dass echte Tokens im Projekt enthalten sind.

## 4.4 Startablauf der Anwendung

Der Start der Anwendung erfolgt über `main.py`. Zuerst wird geprüft, ob der vorbereitete lokale Datensatz und die Datei mit den Mood-Tags vorhanden sind. Falls diese Dateien fehlen, wird die Datenaufbereitung ausgeführt.

Danach wird geprüft, ob das trainierte Modell lokal verfügbar ist. Wenn das Modell nicht vorhanden ist, wird es von Hugging Face heruntergeladen und lokal gespeichert. Anschließend werden die zentralen Komponenten gestartet: Der `MoodPredictor` übernimmt die Mood-Vorhersage, der `MovieRecommender` berechnet die passenden Filme und der `TelegramBot` stellt die Nutzerschnittstelle bereit.

Nach der Initialisierung startet der Bot und wartet auf eingehende Nachrichten. Sobald ein Nutzer eine Nachricht sendet, wird daraus ein Mood-Vektor erzeugt. Dieser wird mit den Filmvektoren verglichen. Die besten Ergebnisse werden anschließend im Chat ausgegeben.

## 4.5 Modularer Aufbau und Einordnung

Der modulare Aufbau trennt die wichtigsten Aufgaben des Systems voneinander. Die Datenaufbereitung ist vom Training getrennt. Das Training ist von der Inferenz getrennt. Die Empfehlungslogik ist wiederum von der Telegram-Schnittstelle getrennt.

Diese Struktur macht das Projekt übersichtlicher und leichter wartbar. Einzelne Bestandteile können angepasst werden, ohne das gesamte System neu aufzubauen. Beispielsweise könnte der Telegram-Bot später durch eine Weboberfläche ersetzt werden. Ebenso könnte das Modell neu trainiert oder der Datensatz erweitert werden, ohne die gesamte Architektur zu verändern.

Insgesamt verbindet die Architektur ein trainiertes Sprachmodell mit einem vektorbasierten Empfehlungssystem. Das Sprachmodell interpretiert die Freitexteingabe und erzeugt einen Mood-Vektor. Der Recommender nutzt diesen Vektor, um passende Filme über einen Ähnlichkeitsvergleich zu finden. Dadurch entsteht ein nachvollziehbarer und erweiterbarer Empfehlungsprozess.
