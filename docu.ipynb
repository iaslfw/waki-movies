```python
{
  "cell_type": "markdown",
  "id": "doku_kapitel_001",
  "metadata": {},
  "source": [
    "# 1. Einleitung und Projektziel\n",
    "\n",
    "## 1.1 Ausgangssituation\n",
    "\n",
    "Filmempfehlungen sind ein typisches Anwendungsfeld für Empfehlungssysteme. Viele bekannte Plattformen schlagen Inhalte auf Basis von Genres, Bewertungen, Nutzerverhalten oder ähnlichen Filmen vor. Solche Ansätze sind grundsätzlich hilfreich, bilden aber nicht immer die tatsächliche Suchabsicht eines Nutzers ab. Häufig sucht ein Nutzer nicht nach einem festen Genre, sondern nach einer bestimmten Stimmung, Atmosphäre oder inhaltlichen Richtung. Eine Anfrage kann zum Beispiel lauten: „Ich möchte einen spannenden Film sehen, der nicht zu düster ist“ oder „Ich suche etwas Lustiges mit Abenteuergefühl“.\n",
    "\n",
    "Klassische Genre-Kategorien sind für solche Suchanfragen nur eingeschränkt geeignet. Ein Film kann mehreren Stimmungen gleichzeitig entsprechen. Ein Science-Fiction-Film kann beispielsweise düster, spannend, emotional oder humorvoll sein. Ebenso kann ein Drama gleichzeitig melancholisch, romantisch und hoffnungsvoll wirken. Für ein Empfehlungssystem ist es daher sinnvoll, Filme nicht nur über starre Kategorien zu betrachten, sondern über feinere Merkmale wie Stimmung, Atmosphäre, Tempo oder thematische Schwerpunkte.\n",
    "\n",
    "## 1.2 Ziel des Projekts\n",
    "\n",
    "Ziel des Projekts „WaKi-Movies“ ist die Entwicklung eines KI-gestützten Empfehlungssystems für Filme. Das System soll eine freie Texteingabe eines Nutzers verarbeiten und daraus passende Filmempfehlungen ableiten. Im Mittelpunkt steht dabei nicht die reine Genre-Suche, sondern die Vorhersage sogenannter Mood-Tags. Diese Mood-Tags beschreiben latente Eigenschaften eines Films, zum Beispiel ob ein Film spannend, lustig, düster, emotional oder abenteuerlich wirkt.\n",
    "\n",
    "Das Projekt kombiniert Methoden aus Natural Language Processing, Deep Learning und Empfehlungssystemen. Eine Nutzereingabe wird zunächst von einem trainierten Sprachmodell analysiert. Anschließend erzeugt das Modell einen Vektor, der die vorhergesagten Stimmungen der Eingabe abbildet. Dieser Nutzervektor wird mit den gespeicherten Vektoren der Filme verglichen. Die Filme mit der höchsten Ähnlichkeit werden anschließend als Empfehlung ausgegeben.\n",
    "\n",
    "## 1.3 Grundidee des Movie Mood-Predictors\n",
    "\n",
    "Die Grundidee besteht aus zwei zentralen Schritten. Im ersten Schritt wird die Freitexteingabe des Nutzers in einen Mood-Vektor umgewandelt. Dafür wird ein Transformer-basiertes Sprachmodell verwendet, das für eine Multi-Label-Klassifikation trainiert wurde. Multi-Label bedeutet in diesem Zusammenhang, dass nicht nur eine einzelne Klasse vorhergesagt wird. Stattdessen können mehrere Stimmungen gleichzeitig erkannt werden.\n",
    "\n",
    "Im zweiten Schritt wird dieser vorhergesagte Mood-Vektor mit den vorhandenen Filmvektoren verglichen. Jeder Film besitzt ebenfalls einen Vektor, der aus den zugrunde liegenden Tag-Daten erzeugt wurde. Für den Vergleich wird die Cosine Similarity verwendet. Dadurch kann gemessen werden, wie ähnlich die Stimmung der Nutzereingabe zur Stimmung eines Films ist. Die Filme mit der höchsten Ähnlichkeit werden als Top-Empfehlungen zurückgegeben.\n",
    "\n",
    "Die Anwendung wird über einen Telegram-Bot nutzbar gemacht. Nutzer müssen daher nicht direkt mit dem Python-Code arbeiten. Stattdessen können sie ihre Filmwünsche als normale Chatnachricht eingeben. Der Bot verarbeitet die Anfrage, ruft die Vorhersage des Modells auf, berechnet passende Empfehlungen und gibt diese anschließend im Chat aus.\n"
  ]
}
```
